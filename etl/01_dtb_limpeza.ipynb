```python
import os
os.chdir(os.path.join('..', 'data', 'dtb'))  # trabalha dentro da pasta de dados
```

```python
#SABER AS COLUNAS E OS TIPOS DE DADOS

for arquivo in [f for f in os.listdir(".") if f.endswith('.xlsx')]:
    print(f"\n{arquivo}:")
    df = pd.read_excel(arquivo, nrows=2)
    for coluna in df.columns:
        valor = df[coluna].iloc[1] if len(df) > 1 else "N/A"
        print(f"  {coluna}: {type(valor).__name__}")
```

```python
# ⬇️ ALTERE AQUI OS NOMES DOS SEUS 3 ARQUIVOS ⬇️
arquivos = ["RELATORIO_DTB_BRASIL_2024_DISTRITOS.xlsx", "RELATORIO_DTB_BRASIL_2024_MUNICIPIOS.xlsx", "RELATORIO_DTB_BRASIL_2024_SUBDISTRITOS.xlsx"]

for arquivo in arquivos:
    try:
        df = pd.read_excel(arquivo)
        df_sem_6_linhas = df.iloc[6:]  # Remove as 6 primeiras linhas
        df_sem_6_linhas.to_excel(arquivo, index=False)
        print(f"✅ {arquivo}: {len(df)} → {len(df_sem_6_linhas)} linhas")
    except Exception as e:
        print(f"❌ {arquivo}: {e}")
```
